# Import Libraries

```python
import pandas as pd
import numpy as np
from faker import Faker
import random
import os
from datetime import datetime, timedelta

fake = Faker("en_IN")

random.seed(42)
np.random.seed(42)
Faker.seed(42)
```

```python
NUM_STUDENTS = 10000
NUM_FACULTY = 250
NUM_DEPARTMENTS = 15
NUM_COURSES = 200
NUM_ATTENDANCE = 500000
NUM_MARKS = 100000
NUM_FEES = 30000
NUM_LIBRARY = 50000
NUM_HOSTEL = 4000
NUM_TRANSPORT = 5000
NUM_LOGIN_LOGS = 200000
NUM_AUDIT_LOGS = 300000

SAVE_PATH = "ERP_Dataset"

os.makedirs(SAVE_PATH, exist_ok=True)
```

```python
departments = [
    "Computer Science",
    "Artificial Intelligence",
    "Information Technology",
    "Mechanical Engineering",
    "Civil Engineering",
    "Electrical Engineering",
    "Electronics",
    "Chemical Engineering",
    "Biotechnology",
    "Mathematics",
    "Physics",
    "Chemistry",
    "MBA",
    "Commerce",
    "Humanities"
]

dept = pd.DataFrame({
    "DepartmentID": range(1, len(departments)+1),
    "DepartmentName": departments
})

dept.to_csv(f"{SAVE_PATH}/departments.csv", index=False)

dept.head()
```

```python
students=[]

for i in range(NUM_STUDENTS):

    students.append({

        "StudentID":100000+i,

        "RollNumber":"RTU"+str(100000+i),

        "FirstName":fake.first_name(),

        "LastName":fake.last_name(),

        "Gender":random.choice(["Male","Female"]),

        "DepartmentID":random.randint(1,15),

        "Semester":random.randint(1,8),

        "CGPA":round(random.uniform(5.5,10),2),

        "Attendance":round(random.uniform(60,100),2),

        "Phone":fake.phone_number(),

        "Email":fake.email(),

        "City":fake.city(),

        "State":fake.state(),

        "AdmissionYear":random.randint(2021,2026)

    })

students=pd.DataFrame(students)

students.to_csv(f"{SAVE_PATH}/students.csv",index=False)

students.head()
```

```python
faculty=[]

designation=["Assistant Professor",
             "Associate Professor",
             "Professor",
             "HOD"]

for i in range(NUM_FACULTY):

    faculty.append({

        "FacultyID":5000+i,

        "FacultyName":fake.name(),

        "DepartmentID":random.randint(1,15),

        "Designation":random.choice(designation),

        "Salary":random.randint(50000,180000),

        "Experience":random.randint(1,35)

    })

faculty=pd.DataFrame(faculty)

faculty.to_csv(f"{SAVE_PATH}/faculty.csv",index=False)
```

```python
courses=[]

for i in range(NUM_COURSES):

    courses.append({

        "CourseID":"CS"+str(1000+i),

        "DepartmentID":random.randint(1,15),

        "FacultyID":random.randint(5000,5249),

        "Credits":random.randint(2,5),

        "Semester":random.randint(1,8)

    })

courses=pd.DataFrame(courses)

courses.to_csv(f"{SAVE_PATH}/courses.csv",index=False)
```

```python
attendance=[]

status=["Present","Absent","Leave"]

for i in range(NUM_ATTENDANCE):

    attendance.append({

        "StudentID":random.randint(100000,109999),

        "CourseID":"CS"+str(random.randint(1000,1199)),

        "Date":fake.date_between("-1y","today"),

        "Status":random.choices(
            status,
            weights=[88,10,2]
        )[0]

    })

attendance=pd.DataFrame(attendance)

attendance.to_csv(f"{SAVE_PATH}/attendance.csv",index=False)
```

```python
marks=[]

for i in range(NUM_MARKS):

    marks.append({

        "StudentID":random.randint(100000,109999),

        "CourseID":"CS"+str(random.randint(1000,1199)),

        "Midterm":random.randint(10,30),

        "Endterm":random.randint(25,70),

        "Assignment":random.randint(5,10)

    })

marks=pd.DataFrame(marks)

marks["Total"]=marks["Midterm"]+marks["Endterm"]+marks["Assignment"]

marks.to_csv(f"{SAVE_PATH}/marks.csv",index=False)
```

```python
fees=[]

status=["Paid","Pending","Partial"]

for i in range(NUM_FEES):

    fees.append({

        "StudentID":random.randint(100000,109999),

        "FeeAmount":random.randint(30000,150000),

        "PaidAmount":random.randint(10000,150000),

        "Status":random.choice(status)

    })

fees=pd.DataFrame(fees)

fees.to_csv(f"{SAVE_PATH}/fees.csv",index=False)
```

```python
logs=[]

for i in range(NUM_LOGIN_LOGS):

    logs.append({

        "UserID":random.randint(100000,109999),

        "LoginTime":fake.date_time_this_year(),

        "IP":fake.ipv4(),

        "Device":random.choice(["Windows","Android","Mac","Linux"]),

        "Browser":random.choice(["Chrome","Firefox","Edge"])

    })

logs=pd.DataFrame(logs)

logs.to_csv(f"{SAVE_PATH}/login_logs.csv",index=False)
```

```python
audit=[]

actions=["Login","Logout","Fee Payment",
         "View Result","Attendance",
         "Library","Profile Update"]

for i in range(NUM_AUDIT_LOGS):

    audit.append({

        "UserID":random.randint(100000,109999),

        "Action":random.choice(actions),

        "Timestamp":fake.date_time_this_year(),

        "Status":"Success"

    })

audit=pd.DataFrame(audit)

audit.to_csv(f"{SAVE_PATH}/audit_logs.csv",index=False)
```
